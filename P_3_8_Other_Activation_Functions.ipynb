# Problem 3.8 - Replacing ReLU w/ Other Activation Functions

```python
import numpy as np
import matplotlib.pyplot as plt

def plot(title, activation_fn):
    mosaic = [
        ['A', 'B', 'C'],
        ['D', 'E', 'F'],
        ['G', 'H', 'I'],
        ['.', 'J', '.']
    ]
    fig, axd = plt.subplot_mosaic(mosaic, layout="constrained", figsize=(9, 9))

    fig.suptitle(title, fontsize="xx-large")

    for ax in axd.values():
        ax.set_xlim(0.0, 4.0)
        ax.set_ylim(-2, 2)
        ax.minorticks_on()
        ax.axhline(0.0, linestyle=":", linewidth=1)

    x = np.linspace(0.0, 4.0, 1000)

    # Pre-activations
    p1 = theta10 + theta11 * x
    p2 = theta20 + theta21 * x
    p3 = theta30 + theta31 * x
    axd["A"].plot(x, p1, linewidth=2, color="blue")
    axd["B"].plot(x, p2, linewidth=2, color="green")
    axd["C"].plot(x, p3, linewidth=2, color="red")

    # Activation
    h1 = activation_fn(p1)
    h2 = activation_fn(p2)
    h3 = activation_fn(p3)
    axd["D"].plot(x, h1, linewidth=2, color="blue")
    axd["E"].plot(x, h2, linewidth=2, color="green")
    axd["F"].plot(x, h3, linewidth=2, color="red")

    # Scaled by phi
    s1 = h1 * phi1
    s2 = h2 * phi2
    s3 = h3 * phi3
    axd["G"].plot(x, s1, linewidth=2, color="blue")
    axd["H"].plot(x, s2, linewidth=2, color="green")
    axd["I"].plot(x, s3, linewidth=2, color="red")

    # Full function
    y = phi0 + s1 + s2 + s3
    axd["J"].plot(x, y, linewidth=2, color="orange")
```

```python
phi0 = -0.23; phi1 = -1.3; phi2 = 1.3; phi3 = 0.66
theta10 = -0.2; theta11 = 0.4; theta20 = -0.9; theta21 = 0.9; theta30 = 1.1; theta31 = -0.7
```

## ReLU

```python
def relu(z: np.ndarray):
    return np.maximum(z, 0)

plot(title="ReLU", activation_fn=relu)
```

## Heaviside Step Function
$$
\text{heaviside}[z] =
\begin{cases}
0 & z < 0 \\
1 & z \ge 0
\end{cases}
$$

```python
def heaviside(z: np.ndarray):
    return np.heaviside(z, 1)

plot(title="Heaviside", activation_fn=heaviside)
```

## Rectangular Function

$$
\text{rect}[z] = 
\begin{cases}
0 & z < 0 \\
1 & 0 \le z \le 1 \\
0 & z > 1
\end{cases}
$$

```python
def rect(z: np.ndarray):
    return np.where((z >= 0) & (z <= 1), 1, 0)

plot(title="Rect", activation_fn=rect)
```

## Hyperbolic Tangent Function

$$
\text{tanh}[z]
$$

```python
plot(title="tanh", activation_fn=np.tanh)
```
