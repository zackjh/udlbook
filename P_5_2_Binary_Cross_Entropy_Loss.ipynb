# Problem 5.2 - Binary Cross Entropy Loss

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def bce_loss(x, y_label):
    return -(1 - y_label) * np.log(1 - x) - y_label * np.log(x)

x = np.linspace(0.00001, 0.99999, 500)
fig, ax = plt.subplots()
ax.set_xlabel(r"$\mathrm{sig}[f[x, \phi]]$")
ax.set_ylabel("loss")
y_0 = bce_loss(x, y_label=0)
y_1 = bce_loss(x, y_label=1)
plt.plot(x, y_0, label="y label = 0")
plt.plot(x, y_1, label="y label = 1")
plt.legend()
plt.show()
```
